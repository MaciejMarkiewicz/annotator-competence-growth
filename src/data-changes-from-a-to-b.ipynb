```python
import pandas as pd

data = pd.read_csv('../data/AB_comparison.csv')
data
```

```python
data['techniques_changed'] = data.apply(
    lambda row: (row['first_question_2_techniques'] != row['repeated_question_2_techniques']) and (row['first_question_2_techniques'] == row['first_question_2_techniques'] or row['repeated_question_2_techniques'] == row['repeated_question_2_techniques']) , axis=1
)

data
```

```python
data.groupby('annotator')['techniques_changed'].mean()
```
