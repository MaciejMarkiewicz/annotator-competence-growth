```python
import pandas as pd

sirsc_1 = pd.read_csv('/content/drive/MyDrive/Dane/Ankieta samooceny kompetencji.csv')
sirsc_2 = pd.read_csv('/content/drive/MyDrive/Dane/Ankieta samooceny kompetencji II.csv')


```

```python
# First measurement
columns_to_process_sirsc_1 = sirsc_1.columns[2:] # Od trzeciej kolumny do końca

for col in columns_to_process_sirsc_1:
    # Konwertuj na string, aby móc użyć metod .str
    sirsc_1[col] = sirsc_1[col].astype(str).str.replace(r'\D', '', regex=True)
    # Konwertuj na typ numeryczny, zmieniając błędy na NaN
    sirsc_1[col] = pd.to_numeric(sirsc_1[col], errors='coerce')

display(sirsc_1.head())
display(sirsc_1[columns_to_process_sirsc_1].dtypes)

# --- Dla DataFrame 'drugi' (odkomentuj i uruchom, jeśli chcesz zastosować to samo do 'drugi') ---
columns_to_process_sirsc_2 = sirsc_2.columns[2:] # Od trzeciej kolumny do końca

for col in columns_to_process_sirsc_2:
    sirsc_2[col] = sirsc_2[col].astype(str).str.replace(r'\D', '', regex=True)
    sirsc_2[col] = pd.to_numeric(sirsc_2[col], errors='coerce')

display(sirsc_2.head())
display(sirsc_2[columns_to_process_sirsc_2].dtypes)
```

Scale recoding

```python
columns_to_recode = [
    'Nie potrafię rozpoznawać technik wpływu społecznego w interakcjach między ludźmi. ',
    'Nie potrafię odróżnić neutralnej komunikacji od prób wpływu.',
    'Nie potrafię wyjaśnić, dlaczego widzę wpływ społeczny w konkretnej sytuacji.'
]

def recode_likert_scale(value):
    if pd.isna(value):
        return value
    # Re-apply 6 - value reversal
    reversed_value = 6 - value
    # Map 0 to 1 to ensure scale is always 1-5 (if 0 is not a valid point)
    return 1 if reversed_value == 0 else reversed_value

for col in columns_to_recode:
    if col in sirsc_1.columns:
        sirsc_1[col] = sirsc_1[col].apply(recode_likert_scale)

display(sirsc_1.head())
```

```python
for col in columns_to_recode:
    if col in sirsc_2.columns:
        sirsc_2[col] = sirsc_2[col].apply(recode_likert_scale)

print("Pierwsze 5 wierszy DataFrame 'drugi' po rekodowaniu:")
display(sirsc_2.head())
```

Subscale scores

```python
# First measurement
columns_perceived_competence = sirsc_1.columns[2:9]
# Pytania od 9 do 12 to kolumny o indeksach od 10 do 13 (łącznie 4 kolumny)
columns_selfconfidence = sirsc_1.columns[10:13]

# Utwórz nową kolumnę 'Poczucie kompetencji' w DataFrame 'pierwszy'
sirsc_1['Perceived_competence'] = sirsc_1[columns_perceived_competence].sum(axis=1)

# Utwórz nową kolumnę 'Poczucie pewności' w DataFrame 'pierwszy'
sirsc_1['Self_confidence'] = sirsc_1[columns_selfconfidence].sum(axis=1)

display(sirsc_1.head())

display(sirsc_1['Perceived_competence'].describe())

display(sirsc_1['Self_confidence'].describe())
```

```python
# Second measurement
columns_perceived_competence = sirsc_2.columns[2:9]
# Pytania od 9 do 12 to kolumny o indeksach od 10 do 13 (łącznie 4 kolumny)
columns_selfconfidence = sirsc_2.columns[10:13]

# Utwórz nową kolumnę 'Poczucie kompetencji' w DataFrame 'pierwszy'
sirsc_2['Perceived_competence'] = sirsc_2[columns_perceived_competence].sum(axis=1)

# Utwórz nową kolumnę 'Poczucie pewności' w DataFrame 'pierwszy'
sirsc_2['Self_confidence'] = sirsc_2[columns_selfconfidence].sum(axis=1)

display(sirsc_2.head())

display(sirsc_2['Perceived_competence'].describe())

display(sirsc_2['Self_confidence'].describe())
```

```python
print("Descriptive statistics for 'Perceived_competence' in sipc_1:")
display(sirsc_1['Perceived_competence'].describe())

print("\nDescriptive statistics for 'Perceived_competence' in sipc_2:")
display(sirsc_2['Perceived_competence'].describe())

print("Descriptive statistics for 'Self-confidence' in sipc_1:")
display(sirsc_1['Self_confidence'].describe())

print("\nDescriptive statistics for 'Self-confidence' in sipc_2:")
display(sirsc_2['Self_confidence'].describe())

```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np # Import numpy for array operations

# Calculate mean scores for 'Perceived_competence' and 'Self_confidence' for sirsc_1
mean_pc_sirsc_1 = sirsc_1['Perceived_competence'].mean()
mean_sc_sirsc_1 = sirsc_1['Self_confidence'].mean()

# Calculate standard deviation scores for 'Perceived_competence' and 'Self_confidence' for sirsc_1
std_pc_sirsc_1 = sirsc_1['Perceived_competence'].std()
std_sc_sirsc_1 = sirsc_1['Self_confidence'].std()

# Calculate mean scores for 'Perceived_competence' and 'Self_confidence' for sirsc_2
mean_pc_sirsc_2 = sirsc_2['Perceived_competence'].mean()
mean_sc_sirsc_2 = sirsc_2['Self_confidence'].mean()

# Calculate standard deviation scores for 'Perceived_competence' and 'Self_confidence' for sirsc_2
std_pc_sirsc_2 = sirsc_2['Perceived_competence'].std()
std_sc_sirsc_2 = sirsc_2['Self_confidence'].std()

# Create a DataFrame for plotting, now including standard deviations
data = {
    'Subscale': ['Perceived Competence', 'Self Confidence', 'Perceived Competence', 'Self Confidence'],
    'Measurement': ['Measurement 1', 'Measurement 1', 'Measurement 2', 'Measurement 2'],
    'Mean Score': [mean_pc_sirsc_1, mean_sc_sirsc_1, mean_pc_sirsc_2, mean_sc_sirsc_2],
    'Std Dev': [std_pc_sirsc_1, std_sc_sirsc_1, std_pc_sirsc_2, std_sc_sirsc_2]
}
df_plot = pd.DataFrame(data)

# Create the bar plot using matplotlib for more direct control over error bars
plt.figure(figsize=(10, 6))

# Define bar width and positions for grouped bars
bar_width = 0.35
r = np.arange(len(df_plot['Subscale'].unique())) # Positions for 'Perceived Competence' and 'Self Confidence'

# Split data by measurement
df_m1 = df_plot[df_plot['Measurement'] == 'Measurement 1']
df_m2 = df_plot[df_plot['Measurement'] == 'Measurement 2']

# Plot bars for Measurement 1
plt.bar(r - bar_width/2, df_m1['Mean Score'], yerr=df_m1['Std Dev'], width=bar_width, capsize=5,
        label='Measurement 1', color='blue', edgecolor='grey')

# Plot bars for Measurement 2
plt.bar(r + bar_width/2, df_m2['Mean Score'], yerr=df_m2['Std Dev'], width=bar_width, capsize=5,
        label='Measurement 2', color='green', edgecolor='grey')


plt.title('Comparison of Mean Subscale Scores Between Two Measurements with Standard Deviations')
plt.ylabel('Mean Score')
plt.xlabel('Subscale')
plt.xticks(r, df_plot['Subscale'].unique()) # Set x-axis labels in the middle of grouped bars
plt.ylim(0, 40)
plt.legend(title='Measurement')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
from scipy import stats

# --- T-test for Perceived Competence ---
print("\n--- Perceived Competence ---")
t_stat_pc, p_value_pc = stats.ttest_ind(sirsc_1['Perceived_competence'], sirsc_2['Perceived_competence'], nan_policy='omit')
print(f"T-statistic (Perceived Competence): {t_stat_pc:.3f}")
print(f"P-value (Perceived Competence): {p_value_pc:.3f}")

if p_value_pc < 0.05:
    print("The difference in 'Perceived Competence' between Measurement 1 and Measurement 2 is statistically significant.")
else:
    print("The difference in 'Perceived Competence' between Measurement 1 and Measurement 2 is NOT statistically significant.")

# --- T-test for Self Confidence ---
print("\n--- Self Confidence ---")
t_stat_sc, p_value_sc = stats.ttest_ind(sirsc_1['Self_confidence'], sirsc_2['Self_confidence'], nan_policy='omit')
print(f"T-statistic (Self Confidence): {t_stat_sc:.3f}")
print(f"P-value (Self Confidence): {p_value_sc:.3f}")

if p_value_sc < 0.05:
    print("The difference in 'Self Confidence' between Measurement 1 and Measurement 2 is statistically significant.")
else:
    print("The difference in 'Self Confidence' between Measurement 1 and Measurement 2 is NOT statistically significant.")
```
